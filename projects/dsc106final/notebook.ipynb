```python
import pandas as pd
```

```python
df = pd.read_csv('data.csv')
df.head()
```

```python
df = df[df['country'] != 'Glabal']
```

```python
avg = df.groupby(['year', 'scenario'])['value'].mean().reset_index()
avg['country'] = 'Global'
avg = avg[['country', 'year', 'scenario', 'value']]
avg
```

```python
df = pd.concat([df, avg]).reset_index(drop=True)
```

```python
df.to_csv('data.csv', index=False)
```
